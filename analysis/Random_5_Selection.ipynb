```python
from collections import defaultdict
from pathlib import Path
import random
import sys

# Works whether the notebook runs from the repository root or from analysis/.
ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "data" / "processed").is_dir())
sys.path.insert(0, str(ROOT))

from analysis.generate import JSONL, paper_text  # paper_text: title and sections, no abstract
from parsing.metadata import load_metadata
from parsing.models import read_jsonl

METADATA_DIR = ROOT / "data"  # metadata_<year>.csv
OUTPUT_FILE = ROOT / "data/random_5_papers_per_year.txt"
PER_YEAR = 5
RANDOM_SEED = 2  # Set an integer for repeatable samples

ids_by_year = defaultdict(list)
for paper_id, row in load_metadata(METADATA_DIR).items():
    ids_by_year[row["year"]].append(paper_id)

papers = {paper.paper_id: paper for paper in read_jsonl(ROOT / JSONL)}

rng = random.Random(RANDOM_SEED)
output_parts = []

for year in sorted(ids_by_year):
    ids = ids_by_year[year]
    for paper_id in rng.sample(ids, min(PER_YEAR, len(ids))):
        output_parts.append(
            f"===== {year} | {paper_id} =====\n\n{paper_text(papers[paper_id])}\n"
        )

result = "\n".join(output_parts)
OUTPUT_FILE.write_text(result, encoding="utf-8")

print(result)
print(f"\nSaved to: {OUTPUT_FILE}")
```
